```python
def request_function(property,specie,page):
    url = (
    "http://aflowlib.duke.edu/search/API/?"
    f"{property}(!null),"
    "compound,"
    "stoichiometry,"
    f"$paging({page}),"
    f"species({specie})")
    
    r = requests.get(url)
    try:
     data = r.json()
    except:
        return pd.DataFrame()
    
    df = pd.DataFrame(data)
    return df

species = ['Cu', 'Ni','Al']
all_data = {}

for specie in species:
    data_specie = pd.DataFrame(columns=['compound','agl_thermal_conductivity_300K','stoichiometry'])

    for i in range(90):
        df = request_function("agl_thermal_conductivity_300K",specie,i)
        
        if df.empty:
            continue
        
        df_v2 = df[["compound","agl_thermal_conductivity_300K","stoichiometry"]]
        data_specie = pd.concat([data_specie,df], ignore_index=True)
    
    all_data[specie] = data_specie

## Save the dictory as a pickle document
with open("data_AFLOW.pkl","wb") as f:
    pickle.dump(all_data,f)
    
```

```python
def read_tab_website (url, no_tab):
    headers = {
    "User-Agent": "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 "
                  "(KHTML, like Gecko) Chrome/120.0 Safari/537.36"
}

    response = requests.get(url, headers=headers)
    response.raise_for_status()

    html = StringIO(response.text)
    tables = pd.read_html(html)

    df = tables[no_tab]
    return df
```

Valence Electrons
Periodic Table Wikipide

```python

url_ec = "https://en.wikipedia.org/wiki/Electron_configurations_of_the_elements_(data_page)"
df_ev = read_tab_website(url_ec,0)
```

```python
def last_numeric(row):
    nums = pd.to_numeric(row, errors="coerce")
    nums = nums.dropna()
    return nums.iloc[-1] if len(nums) > 0 else np.nan 

def extract_valence_electrons(df):
    df_name = df.iloc[1::3] # Extract of name rows
    df_valence = df.iloc[3::3] # Extract of valence rows
    
    df_name = df_name.copy()
    df_name["symbol"] = df_name["Legend"].str.extract(r"\b\d+\s+([A-Z][a-z]?)\b")
    df_name = df_name.reset_index(drop=True)
    
    df_valence = df_valence.copy()
    df_valence.loc[:, "last_value"] = df_valence.apply(last_numeric, axis=1) # Extract the valence electrons
    df_valence = df_valence.reset_index(drop=True)
    
    df_new = pd.DataFrame({
    "symbol": df_name["symbol"],
    "valence": df_valence["last_value"]
})
    return df_new
```

```python
df_valence = extract_valence_electrons(df_ev)
df_valence.to_pickle(path_root+"/Data_Extraction/Valence_electrons.pkl")
```

Atomic Radius (Wikipidia)

```python
url_atr = 'https://en.wikipedia.org/wiki/Atomic_radii_of_the_elements_(data_page)'
df_atr = read_tab_website(url_atr,0)

df_atr = df_atr.dropna(subset=['Metallic'])

df_atomic_radius = df_atr[['symbol','Metallic']]

df_atomic_radius = df_atomic_radius.copy()
df_atomic_radius["Metallic"] = (
    df_atomic_radius["Metallic"]
    .astype(str)
    .str.extract(r"(\d+)")[0]   # extract numeric part
    .astype(float)
)
df_atomic_radius.to_pickle(path_root+"/Data_Extraction/Atomic_radius.pkl")
```

Electronegativity difference

```python
url_electron = 'https://en.wikipedia.org/wiki/Electronegativities_of_the_elements_(data_page)'
df_electron = read_tab_website(url_electron,1)

df_electron['electronegativity'] = pd.to_numeric(df_electron['electronegativity'],errors='coerce')
df_electron = df_electron.dropna(subset=['electronegativity'])

df_electronegativity = df_electron[['Symbol','electronegativity']]
df_electronegativity.to_pickle(path_root+"/Data_Extraction/electronegativity.pkl")
```
